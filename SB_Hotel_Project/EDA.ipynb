```python
import pandas as pd
import numpy as np
import math
import matplotlib.pyplot as plt
import seaborn as sns
import re
```

```python
df = pd.read_excel("/Users/bowiechuang/Downloads/Expedia_hotel_dataset.xlsx")
df = df.drop('hotel_details', axis = 1)
df
```

```python
pattern = r'(Santa Barbara|Goleta|Solvang|Los Olivos|Buellton|Montecito|Carpinteria|Santa Ynez|Los Alamos|Lompoc|Ojai|Ventura)'
df['hotel_city'] = df['hotel_city'].str.extract(pattern, flags = re.IGNORECASE)
df['hotel_city'] = df['hotel_city'].fillna("No City Name")

df['month'] = df['start_date'].dt.month
df['day'] = df['start_date'].dt.day
df['day_of_week'] = df['start_date'].dt.strftime("%A")

#f['Per_Night_Fee'] = df['Per_Night_Fee'].str.replace("\D", '', regex = True)
df['Per_Night_Fee'] = pd.to_numeric(df['Per_Night_Fee'], errors = 'coerce')

df['hotel_rating'] = pd.to_numeric(df['hotel_rating'], errors = 'coerce')

month_to_season = {
    1: 'Winter', 2: 'Winter', 3: 'Spring',
    4: 'Spring', 5: 'Spring', 6: 'Summer',
    7: 'Summer', 8: 'Summer', 9: 'Fall',
    10: 'Fall', 11: 'Fall', 12: 'Winter'
}
df['Season'] = df['month'].map(month_to_season)
df
```
